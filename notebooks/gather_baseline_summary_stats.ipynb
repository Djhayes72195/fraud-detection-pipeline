The purpose of this script is to generate summary statistics that will be used as a baseline for my monitoring service. We are simulating the a scenario where:

- We have 1 day of transaction data including ground truth results (fraud / not fraud).
- The first release of the model and summary statiistics for modeling are based off this data.

```python
from pathlib import Path
import sys
import json
project_root = Path().resolve().parent
if str(project_root) not in sys.path:
    sys.path.insert(0, str(project_root))
import pandas as pd
from data_processing.data_ingestion import load_days

baseline_stats_path = Path("C:\Users\Djhay\OneDrive\Desktop\Projects\fraud-detection-pipeline\logs")
```

```python
df = load_days(start_day=1, end_day=1)
df = df.drop(columns=["sid", "Day", "TransactionID", "PerturbationScheme", "dt"])
```

```python
desc = df.describe().T
baseline_stats = desc[["mean", "std", "min", "25%", "50%", "75%", "max", "count"]]
baseline_stats = baseline_stats.rename(columns={
    "25%": "q25", "50%": "q50", "75%": "q75"
}).to_dict(orient="index")
```

```python
with open(baseline_stats_path, "w") as f:
    json.dump(baseline_stats, f)
```
